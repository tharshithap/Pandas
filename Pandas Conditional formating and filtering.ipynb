# Conditional Formatting or Filtering

# Syntax
- df[col]-condition - boolean df
- df[df[col]-condition] - filter the df

```python
import pandas as pd
```

```python
names = ['Ram', 'John', 'Abdul','Siva', 'Adam', 'Rahman']
maths =  [45, 40, 60, 80, 90, 75]
english =[60,75,90,65,80,60]
di = {"stu_names":names,"Maths_marks":maths,'English_marks':english}
s_df=pd.DataFrame(di)
s_df
```

```python
s_df['Maths_marks']>=60
```

```python
s_df[s_df['Maths_marks']>=60]
```

```python
s_df[s_df['Maths_marks']>=60][["stu_names","Maths_marks"]]
```

# sorting the values in data frames

```python
s_df[s_df['Maths_marks']>=60][["stu_names","Maths_marks"]].sort_values("Maths_marks")
```

# Filtering the data based on condition 

# & condition 

```python
s_df[(s_df['Maths_marks']>=60) & (s_df['English_marks']>=70)]
```

# or condition

```python
s_df[(s_df['Maths_marks']>=60) | (s_df['English_marks']>=70)]
```

```python
s_df[~((s_df['Maths_marks']>=60))]
```

# ^ XOR condition 

```python
s_df[(s_df['Maths_marks']>=60) ^ (s_df['English_marks']>=70)]
```

```python
s_df[s_df["stu_names"]!="Jhon"]
```

```python
s_df[s_df["stu_names"]=="John"]
```

# BI (Business Intelligence) section – types of graphs, these are the most commonly used:

- Bar Chart – Compare values between categories.
- Line Chart – Show trends over time.
- Pie Chart – Show percentage/share of a total.
- Column Chart – Compare categories using vertical bars.
- Scatter Plot – Show the relationship between two variables.
- Area Chart – Show trends and magnitude over time.
- Histogram – Show the distribution of numerical data.
- Funnel Chart – Show stages in a process, such as leads → customers.
- Map Chart – Show data based on geographical locations.
- KPI Card – Display an important number such as revenue, sales, or profit.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

data = {
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR",
                   "Sales", "Sales", "Sales", "Marketing", "Marketing", "Marketing"],
    
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male", "Female"],
    
    "Age": [23, 25, 28, 30, 27, 32, 24, 29, 35, 26, 31, 34],
    
    "Salary": [30000, 35000, 40000, 32000, 36000, 42000,
               28000, 33000, 45000, 31000, 39000, 48000],
    
    "Experience": [1, 2, 4, 5, 3, 7, 1, 4, 8, 2, 6, 9]
}

df = pd.DataFrame(data)

df
```

# Strip Plot

```python
sns.stripplot(x="Department", y="Salary", data=df)

plt.title("Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()
```

# Strip Plot with jitter

```python
sns.stripplot(
    x="Department",
    y="Salary",
    data=df,
    jitter=True
)

plt.title("Salary by Department")
plt.show()
```

# Swarm Plot

```python
sns.swarmplot(
    x="Department",
    y="Salary",
    data=df
)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()
```

# 6. Cross Plot / Crosstab

If by cross plot you mean a cross-tabulation, it is used to compare two categorical variables.

```python
cross_table = pd.crosstab(
    df["Department"],
    df["Gender"]
)

print(cross_table)
```

```python
cross_table.plot(kind="bar")

plt.title("Department vs Gender")
plt.xlabel("Department")
plt.ylabel("Count")

plt.show()
```

# Category vs Category

#  Stacked Chart

- A stacked chart puts multiple values on top of each other.

- Let's count male/female employees in each department.

```python
# With: stacked=True
cross_table = pd.crosstab(
    df["Department"],
    df["Gender"]
)

cross_table.plot(
    kind="bar",
    stacked=True
)

plt.title("Gender Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# with stack = False ------>   the bars appear side-by-side.
cross_table = pd.crosstab(
    df["Department"],
    df["Gender"]
)

cross_table.plot(
    kind="bar",
    stacked= False
)

plt.title("Gender Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()

```

# For a proper count-based stacked chart:

```python


count_data = (
    df.groupby(["Department", "Gender"])
      .size()
      .reset_index(name="Count")
)

fig = px.bar(
    count_data,
    x="Department",
    y="Count",
    color="Gender",
    title="Gender Count by Department"
)

fig.show()
```
