# 🟠 Level 3 – New Columns & Calculations

Create a new column called Total containing the total marks of Python, SQL, and Pandas.
    
Create an Average column.
    
Find the student with the highest average.
    
Find the student with the lowest average.
    
Create a Result column:
    
Average >= 40 → "Pass"
    
Average < 40 → "Fail"
    
Create a Performance column:
    
Average >= 85 → "Excellent"
    
Average >= 75 → "Good"
    
Average >= 60 → "Average"
    
Otherwise → "Needs Improvement"
    
Create a column Attendance_Status:
    
Attendance >= 75 → "Eligible"
    
Attendance < 75 → "Not Eligible"


```python
df["Total"] = df['Python']+df['SQL']+df['Pandas']
df
```

```python
df['Average'] = df["Total"]/3
df
```

```python
df.loc[df[['Average']].idxmax()]
```

```python
df.loc[df[['Average']].idxmin()]
```

```python
df["Result"] = "Fail"

df.loc[df["Average"] >= 40, "Result"] = "Pass"

df
```

```python
df["Performance"] = "Needs Improvement"

df.loc[df["Average"] >= 60, "Performance"] = "Average"
df.loc[df["Average"] >= 75, "Performance"] = "Good"
df.loc[df["Average"] >= 85, "Performance"] = "Excellent"

df
```

```python
df["Attendance_Status"] = "Not Eligible"

df.loc[df["Attendance"] >= 75, "Attendance_Status"] = "Eligible"

df
```

```python
df.drop("Attendance_Status", axis=1, inplace=True)
```
