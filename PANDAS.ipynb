#  Data Frame Indexing and Slicing

```python
score_df.iloc[[0,3,5],[1]]
```

```python
score_df.iloc[[0,3,5],[1,2]]
```

# slicing

```python
score_df.iloc[2:,0:3:1]
```

```python
names = ['Ram', 'John', 'Abdul','Siva', 'Adam', 'Rahman']
maths =  [45, 40, 60, 80, 90, 75]
english =[60,75,90,65,80,60]
di = {"stu_names":names,"Maths_marks":maths,'English_marks':english}
s_df=pd.DataFrame(di)
s_df
```

```python
# index 1 rows and maths using iloc
s_df.iloc[[0],[1]]
```

```python
# index 1,2,3 and maths mark using iloc
s_df.iloc[[1,2,3],[1]]
```

```python
# index 1,2,3 and english mark using iloc
s_df.iloc[[1,3,4],[2]]
```
